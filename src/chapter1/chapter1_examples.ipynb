## 5. Object-oriented programming with classes

This chapter introduces JavaScript classes, encapsulation, inheritance, and polymorphism.

````javascript
(() => {
  globalThis.Book = class Book {
    #percentagePerSale = 0.12;

    constructor(title, pages, isbn) {
      this.title = title;
      this.pages = pages;
      this.isbn = isbn;
    }

    get price() {
      return this.pages * this.#percentagePerSale;
    }

    static copiesSold = 0;

    static sellCopy() {
      this.copiesSold++;
    }

    printIsbn() {
      console.log(this.isbn);
    }
  };

  const myBook = new globalThis.Book('title', 400, 'isbn');
  console.log('myBook.title:', myBook.title);
  myBook.title = 'new title';
  console.log('updated title:', myBook.title);
  console.log('price:', myBook.price);
  console.log('copiesSold before:', globalThis.Book.copiesSold);
  globalThis.Book.sellCopy();
  console.log('copiesSold after:', globalThis.Book.copiesSold);

  globalThis.Ebook = class Ebook extends globalThis.Book {
    constructor(title, pages, isbn, format) {
      super(title, pages, isbn);
      this.format = format;
    }

    printIsbn() {
      console.log('Ebook ISBN:', this.isbn);
    }
  };

  const myEbook = new globalThis.Ebook('DS Ebook', 401, 'isbn 123', 'pdf');
  myBook.printIsbn();
  myEbook.printIsbn();
})();
````

## 7. Prototypal inheritance in JavaScript

In JavaScript, functions play a central role in creating and managing objects. JavaScript is prototype-based at its core (classes are syntactic sugar over prototypes).

### Functions as object factories

A factory function creates and returns objects:

```javascript
function createCar(make, model) {
  return {
    make,
    model,
    drive() {
      console.log(`Driving a ${this.make} ${this.model}`);
    }
  };
}

const car1 = createCar('Toyota', 'Corolla');
car1.drive(); // Driving a Toyota Corolla
```

### Constructor functions

Before ES6 classes, constructor functions were the standard way to create many instances:

```javascript
function CarCtor(make, model) {
  this.make = make;
  this.model = model;
  this.drive = function () {
    console.log(`Driving a ${this.make} ${this.model}`);
  };
}

const car2 = new CarCtor('Honda', 'Civic');
car2.drive(); // Driving a Honda Civic
```

When `new CarCtor(...)` runs, JavaScript:
1. Creates a new object (`{}`).
2. Sets its prototype to `CarCtor.prototype`.
3. Binds `this` inside `CarCtor` to the new object.
4. Returns the new object (unless something else is explicitly returned).

````javascript
// Constructor function + shared prototype method
function Car(make, model) {
  this.make = make;
  this.model = model;
}

Car.prototype.drive = function () {
  console.log(`Driving a ${this.make} ${this.model}`);
};

globalThis.chapter1_car3 = new Car('Ford', 'Mustang');
globalThis.chapter1_car3.drive(); // Driving a Ford Mustang

// ES6 class syntax (sugar over constructor + prototype)
globalThis.Chapter1CarClass = class Chapter1CarClass {
  constructor(make, model) {
    this.make = make;
    this.model = model;
  }

  drive() {
    console.log(`Driving a ${this.make} ${this.model}`);
  }
};

globalThis.chapter1_car4 = new globalThis.Chapter1CarClass('Tesla', 'Model 3');
globalThis.chapter1_car4.drive(); // Driving a Tesla Model 3
````

## 8. OOP concepts in JavaScript

Object-oriented programming in JavaScript is usually explained through five concepts:

1. Objects: entities that bundle state (properties) and behavior (methods).
2. Classes: blueprints that define how objects are created.
3. Encapsulation: grouping data and methods together, with controlled access.
4. Inheritance: child classes reusing and extending parent behavior.
5. Polymorphism: same method name, different behavior depending on the object.

### Book class recap

In our `Book` example:
- `title`, `pages`, and `isbn` are public properties.
- `#percentagePerSale` is private and only accessible inside the class.
- `get price()` computes a value based on internal state.
- `static copiesSold` and `static sellCopy()` are class-level members.

### Inheritance and polymorphism recap

`Ebook extends Book` to reuse constructor behavior via `super(...)` and then override `printIsbn()`. Calling `printIsbn()` on `Book` and `Ebook` instances demonstrates polymorphism because each object responds in its own way.

````javascript
(() => {
  // Inheritance + polymorphism example with class static members
  console.log('Book.copiesSold:', globalThis.Book.copiesSold);
  globalThis.Book.sellCopy();
  globalThis.Book.sellCopy();
  console.log('Book.copiesSold after 2 sales:', globalThis.Book.copiesSold);

  globalThis.Ebook.sellCopy();
  console.log('Ebook.copiesSold after Ebook.sellCopy():', globalThis.Ebook.copiesSold);

  const polyBook = new globalThis.Book('title', 400, 'isbn');
  const polyEbook = new globalThis.Ebook('DS Ebook', 401, 'isbn 123', 'pdf');
  polyBook.printIsbn();
  polyEbook.printIsbn();
})();
````

## 9. Modern JavaScript techniques

### Arrow functions

Arrow functions provide a shorter syntax for function expressions and are useful for concise algorithm code.

### Spread and rest operators

- Spread (`...`) expands arrays or iterables into arguments.
- Rest (`...`) collects remaining arguments into an array.

### Exponentiation operator

The exponentiation operator (`**`) is a readable way to express powers and is useful in math-heavy code (for example, geometry or complexity-related calculations).

````javascript
(() => {
  const circleAreaFn = function (radius) {
    const PI = 3.14;
    const area = PI * radius * radius;
    return area;
  };
  console.log('circleAreaFn(2):', circleAreaFn(2)); // 12.56

  const circleArea = (radius) => {
    const PI = 3.14;
    return PI * radius * radius;
  };
  console.log('circleArea(2):', circleArea(2)); // 12.56

  const circleAreaSimp = (radius) => 3.14 * radius * radius;
  console.log('circleAreaSimp(2):', circleAreaSimp(2)); // 12.56

  const hello = () => console.log('hello!');
  hello();

  const sumNumbers = (x, y, z) => x + y + z;
  const numbers = [1, 2, 3];
  console.log('sum(...numbers):', sumNumbers(...numbers)); // 6
  console.log('sum.apply(null, numbers):', sumNumbers.apply(null, numbers)); // 6

  const restParameterFunction = (x, y, ...a) => (x + y) * a.length;
  console.log('restParameterFunction(1, 2, "hello", true, 7):', restParameterFunction(1, 2, 'hello', true, 7)); // 9

  const radius = 2;
  let area = 3.14 * radius * radius;
  console.log('area with multiplication:', area);
  area = 3.14 * Math.pow(radius, 2);
  console.log('area with Math.pow:', area);
  area = 3.14 * (radius ** 2);
  console.log('area with exponentiation operator:', area);
})();
````

## 10. TypeScript ideas in JavaScript

TypeScript adds compile-time type checking on top of JavaScript. In this notebook, we keep the examples runnable in JavaScript and show the closest JavaScript equivalents.

### Type inference style examples

JavaScript does not need type annotations, but you can still write clear, typed-by-convention code:

```javascript
const age = 20;
const existsFlag = true;
const language = 'JavaScript';

let favoriteLanguage;
const langs = ['JavaScript', 'Ruby', 'Python'];
favoriteLanguage = langs[0];
```

### Interfaces as object-shape checks

JavaScript does not have TypeScript interfaces, but you can document object shapes with JSDoc and use runtime checks when needed:

```javascript
/**
 * @typedef {Object} Person
 * @property {string} name
 * @property {number} age
 */

function printName(person) {
  console.log(person.name);
}

const john = { name: 'John', age: 21 };
const mary = { name: 'Mary', age: 21, phone: '123-45678' };
printName(john);
printName(mary);
```

### Comparable behavior with plain JavaScript

Instead of a TypeScript interface, JavaScript code can rely on a shared method contract:

```javascript
class MyObject {
  constructor(age) {
    this.age = age;
  }

  compareTo(b) {
    if (this.age === b.age) {
      return 0;
    }
    return this.age > b.age ? 1 : -1;
  }
}

function compareTwoObjects(a, b) {
  console.log(a.compareTo(b));
  console.log(b.compareTo(a));
}

const obj1 = new MyObject(10);
const obj2 = new MyObject(20);
compareTwoObjects(obj1, obj2);
```

This is the JavaScript version of the polymorphism example: any object with a compatible `compareTo` method can participate.

### Generic-style reusable behavior

TypeScript generics let you write reusable code that keeps the expected input type aligned. In JavaScript, the same idea is usually expressed through a shared method contract and careful naming:

```javascript
function compareTwoValues(a, b) {
  return a.compareTo(b);
}
```

The function above is reusable because it works with any object that exposes the expected behavior, even though JavaScript does not have a generic type parameter like `T`.

### Enums with frozen objects

JavaScript does not have enums, but you can create a frozen lookup object:

```javascript
const Compare = Object.freeze({
  LESS_THAN: -1,
  BIGGER_THAN: 1,
  EQUALS: 0
});
```

### Type aliases with naming conventions

JavaScript does not have TypeScript type aliases, but you can still name shared shapes clearly:

```javascript
/**
 * @typedef {Object} User
 * @property {string} id
 * @property {string} name
 */

const user = {
  id: 'user-1',
  name: 'Packt'
};
```

### Optional type checking in JavaScript files

You can enable editor-assisted checking in JavaScript with:

```javascript
// @ts-check
```

Combined with JSDoc annotations, this gives type-aware feedback in editors while keeping the file valid JavaScript. For example:

```javascript
/**
 * @param {number} radius
 * @returns {number}
 */
function circleAreaFn(radius) {
  const PI = 3.14;
  return PI * radius * radius;
}
```

If you pass the wrong kind of value to `circleAreaFn`, the editor can flag it when `// @ts-check` is enabled.

````javascript
(() => {
  class TypeScriptIdeaComparable {
    constructor(age) {
      this.age = age;
    }

    compareTo(b) {
      if (this.age === b.age) {
        return 0;
      }
      return this.age > b.age ? 1 : -1;
    }
  }

  function compareTwoObjects(a, b) {
    console.log(a.compareTo(b));
    console.log(b.compareTo(a));
  }

  function compareTwoValues(a, b) {
    return a.compareTo(b);
  }

  const obj1 = new TypeScriptIdeaComparable(10);
  const obj2 = new TypeScriptIdeaComparable(20);
  compareTwoObjects(obj1, obj2);
  console.log('compareTwoValues(obj1, obj2):', compareTwoValues(obj1, obj2));

  const Compare = Object.freeze({
    LESS_THAN: -1,
    BIGGER_THAN: 1,
    EQUALS: 0
  });

  function compareTo(a, b) {
    if (a.age === b.age) {
      return Compare.EQUALS;
    }
    return a.age > b.age ? Compare.BIGGER_THAN : Compare.LESS_THAN;
  }

  console.log('compareTo(obj1, obj2):', compareTo(obj1, obj2));

  /**
   * @param {number} radius
   * @returns {number}
   */
  function circleAreaFn(radius) {
    const PI = 3.14;
    return PI * radius * radius;
  }

  console.log('circleAreaFn(2):', circleAreaFn(2));
})();


````
