## 1. Why Big O matters

Big O notation describes how runtime or memory usage scales as input size grows. It helps you compare algorithms independent of hardware speed.

````javascript
(() => {
  function linearSearch(arr, target) {
    let checks = 0;
    for (let i = 0; i < arr.length; i++) {
      checks++;
      if (arr[i] === target) return { index: i, checks };
    }
    return { index: -1, checks };
  }

  const small = Array.from({ length: 10 }, (_, i) => i + 1);
  const large = Array.from({ length: 100 }, (_, i) => i + 1);
  console.log('search in n=10:', linearSearch(small, 10));
  console.log('search in n=100:', linearSearch(large, 100));
})();
````

## 2. Common time complexities

- `O(1)` constant
- `O(log n)` logarithmic
- `O(n)` linear
- `O(n^2)` quadratic
- `O(2^n)` exponential
- `O(n!)` factorial

````javascript
(() => {
  // O(1)
  function getFirstElement(arr) {
    return arr[0];
  }

  // O(log n)
  function binarySearch(sortedArr, target) {
    let left = 0;
    let right = sortedArr.length - 1;
    let steps = 0;
    while (left <= right) {
      steps++;
      const mid = Math.floor((left + right) / 2);
      if (sortedArr[mid] === target) return { index: mid, steps };
      if (sortedArr[mid] < target) left = mid + 1;
      else right = mid - 1;
    }
    return { index: -1, steps };
  }

  // O(n)
  function findMax(arr) {
    let max = -Infinity;
    for (let i = 0; i < arr.length; i++) {
      if (arr[i] > max) max = arr[i];
    }
    return max;
  }

  // O(n^2)
  function hasDuplicates(arr) {
    for (let i = 0; i < arr.length; i++) {
      for (let j = i + 1; j < arr.length; j++) {
        if (arr[i] === arr[j]) return true;
      }
    }
    return false;
  }

  const data = [3, 1, 4, 7, 2, 9, 5];
  const sorted = [...data].sort((a, b) => a - b);
  console.log('O(1) getFirstElement:', getFirstElement(data));
  console.log('O(log n) binarySearch:', binarySearch(sorted, 7));
  console.log('O(n) findMax:', findMax(data));
  console.log('O(n^2) hasDuplicates:', hasDuplicates([1, 2, 3, 2]));
})();
````

## 3. Exponential and factorial growth

These growth rates explode quickly and are usually impractical for medium or large inputs.

````javascript
(() => {
  // O(2^n): recursive Fibonacci without memoization
  function fib(n) {
    if (n <= 1) return n;
    return fib(n - 1) + fib(n - 2);
  }

  // O(n!): count permutations recursively
  function permutationCount(n) {
    if (n <= 1) return 1;
    return n * permutationCount(n - 1);
  }

  console.log('fib(10):', fib(10));
  console.log('10! permutations:', permutationCount(10));
})();
````

The table below is generated in code so you can inspect how operation counts scale.

## 5. Space complexity

Space complexity measures extra memory used by an algorithm beyond the input itself.

````javascript
(() => {
  // O(1) auxiliary space
  function sumInPlace(arr) {
    let total = 0;
    for (let i = 0; i < arr.length; i++) total += arr[i];
    return total;
  }

  // O(n) auxiliary space
  function copyArray(arr) {
    const copy = [];
    for (let i = 0; i < arr.length; i++) copy.push(arr[i]);
    return copy;
  }

  const sample = [1, 2, 3, 4, 5];
  console.log('O(1) sumInPlace:', sumInPlace(sample));
  console.log('O(n) copyArray:', copyArray(sample));
})();
````

## 6. Calculating complexity step-by-step

Analyze operations, loops, nested loops, worst-case behavior, and dominant terms.

````javascript
(() => {
  function createPairs(arr) {
    const pairs = [];
    let pushes = 0;

    for (let i = 0; i < arr.length; i++) {
      for (let j = 0; j < arr.length; j++) {
        pairs.push([arr[i], arr[j]]);
        pushes++;
      }
    }

    return { pairs, pushes };
  }

  const arr = [1, 2, 3, 4];
  const result = createPairs(arr);
  console.log('pairs length:', result.pairs.length);
  console.log('push operations:', result.pushes);
  console.log('time complexity: O(n^2), space complexity: O(n^2)');
})();
````

## 7. Big O in interviews

In interviews, communicate: (1) current complexity, (2) bottlenecks, (3) tradeoffs, and (4) a better approach if possible.

````javascript
(() => {
  // Baseline O(n^2) common-element check
  function hasCommonElementsQuadratic(a, b) {
    for (let i = 0; i < a.length; i++) {
      for (let j = 0; j < b.length; j++) {
        if (a[i] === b[j]) return true;
      }
    }
    return false;
  }

  // Optimized average O(n + m) using a Set
  function hasCommonElementsLinear(a, b) {
    const seen = new Set(a);
    for (const value of b) {
      if (seen.has(value)) return true;
    }
    return false;
  }

  const a = [1, 3, 5, 7, 9];
  const b = [2, 4, 6, 8, 9];
  console.log('quadratic:', hasCommonElementsQuadratic(a, b));
  console.log('linear with Set:', hasCommonElementsLinear(a, b));
})();
````

## 8. Exercises

Try to determine time and space complexity for each function before reading outputs.

````javascript
(() => {
  const isEven = (array) => array.length % 2 === 0;

  function calculateSum(array) {
    let sum = 0;
    for (let i = 0; i < array.length; i++) {
      sum += array[i];
    }
    return sum;
  }

  function hasCommonElements(array1, array2) {
    for (let i = 0; i < array1.length; i++) {
      for (let j = 0; j < array2.length; j++) {
        if (array1[i] === array2[j]) return true;
      }
    }
    return false;
  }

  function getOddNumbers(array) {
    const oddNumbers = [];
    for (let i = 0; i < array.length; i++) {
      if (array[i] % 2 !== 0) {
        oddNumbers.push(array[i]);
      }
    }
    return oddNumbers;
  }

  const demo = [1, 2, 3, 4, 5, 6, 7];
  console.log('isEven:', isEven(demo));
  console.log('calculateSum:', calculateSum(demo));
  console.log('hasCommonElements:', hasCommonElements([1, 3, 5], [2, 4, 5]));
  console.log('getOddNumbers:', getOddNumbers(demo));
})();
````

## 9. Exercise answers

- `isEven`: time `O(1)`, space `O(1)`
- `calculateSum`: time `O(n)`, space `O(1)`
- `hasCommonElements`: time `O(n*m)` (or `O(n^2)` when same-size arrays), space `O(1)`
- `getOddNumbers`: time `O(n)`, space `O(n)` in the worst case
